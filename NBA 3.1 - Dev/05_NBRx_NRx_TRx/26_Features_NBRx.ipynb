## Create Features

```python
index_cols = ["PROVIDER_ID", "SVC_DT_M"]
```

###1.1 Product Group

```python
# Remove null values from Drug class column
df_category_claims = (
  hcp_lrx_claims
  .withColumn("PRODUCT_GROUP",
                F.when(((col("PRODUCT_GROUP").isNull()) | (col('PRODUCT_GROUP') == 'NULL')), 'UNKNOWN_PG').otherwise(col('PRODUCT_GROUP')))
  # .filter("DAYS_SUPPLY <> 'ignore'")  
)

# Drug category
df_category = (
  df_category_claims
  .groupBy(index_cols).pivot("PRODUCT_GROUP")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)

# clean column name
df_category = df_category.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_PRODUCT_GROUP_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_category.columns])
```

###1.2 Specialty Group

```python
# Remove null values from Drug class column
df_specialty_grp_claims = (
  hcp_lrx_claims
  .withColumn("SPECIALTY_GROUP",
                F.when(((col("SPECIALTY_GROUP").isNull()) | (col('SPECIALTY_GROUP') == 'NULL')), 'UNKNOWN').otherwise(col('SPECIALTY_GROUP')))
  # .filter("DAYS_SUPPLY <> 'ignore'")  
)

# Drug category
df_specialty = (
  df_specialty_grp_claims
  .groupBy(index_cols).pivot("SPECIALTY_GROUP")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)

# clean column name
df_specialty = df_specialty.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_SPEC_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_specialty.columns])
```

###1.3 Switch From

```python
# Remove null values from Drug class column
df_switch_frm_claims = (
  hcp_lrx_claims
  .withColumn("SWITCH_FROM",
                F.when(((col("SWITCH_FROM").isNull()) | (col('SWITCH_FROM') == 'NULL')), 'UNKNOWN').otherwise(col('SWITCH_FROM')))
  # .filter("DAYS_SUPPLY <> 'ignore'")  
)

# Drug category
df_switch = (
  df_switch_frm_claims
  .groupBy(index_cols).pivot("SWITCH_FROM")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)

# clean column name
df_switch = df_switch.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_SWITCH_FROM_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_switch.columns])
```

###1.4 Add on to

```python
# Remove null values from Drug class column
df_add_on_claims = (
  hcp_lrx_claims
  .withColumn("ADD_ON_TO",
                F.when(((col("ADD_ON_TO").isNull()) | (col('ADD_ON_TO') == 'NULL')), 'UNKNOWN').otherwise(col('ADD_ON_TO')))
  # .filter("DAYS_SUPPLY <> 'ignore'")  
)

# Drug category
df_add_on_to = (
  df_add_on_claims
  .groupBy(index_cols).pivot("ADD_ON_TO")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)

# clean column name
df_add_on_to = df_add_on_to.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_ADD_ON_TO_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_add_on_to.columns])
```

###1.5 Age Group

```python
# age group
df_patient_age = (
  hcp_lrx_claims
  .groupBy(index_cols).pivot("AGE_GROUP")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )

  #.na.fill(value=0)
)

# clean column name
df_patient_age = df_patient_age.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_AGE_GROUP_' + c.replace(", ", " ").replace("+", "above").replace("/", " or ").replace("-", "_").replace("<", "LESS_THAN_").replace("<", "GREATER_THAN_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_patient_age.columns])

### Change col name 
```

###1.6 Pay Type

```python
# Join the dataframes 
df_payer = (
  hcp_lrx_claims
  .groupBy(index_cols).pivot("PAY_TYPE")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)


# clean column name
df_payer = df_payer.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_PAY_TYPE_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_payer.columns])
```

###1.7 Source Of Business

```python
# Drug brand
df_sob = (
  hcp_lrx_claims
  .groupBy(index_cols).pivot("SOURCE_OF_BUSINESS")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)

# clean column name
df_sob = df_sob.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_SOB_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_sob.columns])
```

###1.8 Channel Level

```python
# Drug brand
df_channels = (
  hcp_lrx_claims
  .groupBy(index_cols).pivot("RX_CHANNEL")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)

# clean column name
df_channels = df_channels.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_CHANNEL_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_channels.columns])
```

###1.9 Indication Group

```python
# Join the dataframes 
df_indication_grp = (
  hcp_lrx_claims
  .groupBy(index_cols).pivot("INDICATION_GROUP")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
  #.na.fill(value=0)
)

# clean column name
df_indication_grp = df_indication_grp.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_INDICATION_GROUP_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_indication_grp.columns])
```

###2.0 Hypertension Flag

```python
# Remove null values from hypertension column
df_hypertension_claims = (
  hcp_lrx_claims
  .withColumn("HYPERTENSION_FLAG",
                F.when(((col("HYPERTENSION_FLAG").isNull()) | (col('HYPERTENSION_FLAG') == 'NULL')), 'UNKNOWN_HF').otherwise(col('HYPERTENSION_FLAG')))
  # .filter("DAYS_SUPPLY <> 'ignore'")  
)

df_hypertension_flag = (
  df_hypertension_claims
  .groupBy(index_cols).pivot("HYPERTENSION_FLAG")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )

  #.na.fill(value=0)
)

# clean column name
df_hypertension_flag = df_hypertension_flag.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_HYPERTENSION_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").replace("&", "").upper()) if c not in index_cols else col(c) for c in df_hypertension_flag.columns])
```

###2.1 Gender

```python
# Remove null values from column
df_gender_claims = (
  hcp_lrx_claims
  .withColumn("GENDER",
                F.when(((col("GENDER").isNull()) | (col('GENDER') == 'NULL')), 'UNKNOWN').otherwise(col('GENDER')))
  # .filter("DAYS_SUPPLY <> 'ignore'")  
)

# gender category
df_gender = (
  df_gender_claims
  .groupBy(index_cols).pivot("GENDER")
  .agg(
    F.sum("NBRX").alias("SUM_NBRX")
    )
 #.na.fill(value=0)
)

# clean column name
df_gender = df_gender.select([F.col(c).alias('TOTAL_HCP_NBRX_CLAIMS_GENDER_' + c.replace(", ", " ").replace("+", "and").replace("/", " or ").replace("-", "_").replace(" ", "_").replace("__", "_").replace("(", "").replace(")", "").upper()) if c not in index_cols else col(c) for c in df_gender.columns])
```

```python
# combine features
df_spine = (
  df_category
  .join(df_specialty, how = 'left', on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_switch, how = "left", on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_add_on_to, how = "left", on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_patient_age, how = 'left', on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_payer, how = 'left', on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_sob, how = 'left', on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_channels, how = "left", on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_indication_grp, how = "left", on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_hypertension_flag, how = 'left', on = ["PROVIDER_ID", "SVC_DT_M"])
  .join(df_gender, how = "left", on = ["PROVIDER_ID", "SVC_DT_M"]))
```

```python
del df_specialty, df_switch, df_add_on_to, df_patient_age, df_payer, df_sob, df_channels, df_indication_grp, df_hypertension_flag,df_gender #, df_pri_specialty
gc.collect()
```
