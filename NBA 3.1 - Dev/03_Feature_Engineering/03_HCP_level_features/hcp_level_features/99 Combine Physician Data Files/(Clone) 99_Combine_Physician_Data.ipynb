### Specialty - One hot encoding

```python
# Filter out rows where Specialty is null before pivoting
# Do NOT fillna — we want rows with null state to stay null across one-hot columns
df_unencoded_cols = df_unencoded_cols.filter(F.col("SPECIALTY_GROUP").isNotNull())

# Assign 1 to specialty_presence only for valid rows
df_unencoded_cols = df_unencoded_cols.withColumn("specialty_presence", lit(1))

# Pivot the valid records to one-hot encode SPECIALTY_GEO
specialty_features = df_unencoded_cols.groupBy("HCP_COHORT_ID").pivot("SPECIALTY_GROUP").agg(F.count("specialty_presence"))

# Replace nulls with 0
specialty_cols = [c for c in specialty_features.columns if c not in ["HCP_COHORT_ID"]]
print("# Specialty Columns :", len(specialty_cols))
specialty_features = specialty_features.fillna(0, subset = specialty_cols)
```
