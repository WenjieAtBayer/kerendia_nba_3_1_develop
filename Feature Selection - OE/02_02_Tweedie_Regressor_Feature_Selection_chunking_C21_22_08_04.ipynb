```python
def feature_selection(df_data,features_list):

  # Divide the data into x and y
  target_df = df_data['TARGET']
  features_list.remove("TARGET")
  X = df_data.drop('TARGET', axis =1)

  #Standard deviation
  x_std = X.std(axis = 0, skipna = True)
  x_std = x_std.tolist()
  y_std = target_df.std(axis = 0, skipna = True)

  #Fit the data
  model = linear_model.TweedieRegressor(alpha=1000)
  model = model.fit(X, df_data['TARGET'])

  #Feature Importance
  coeff = model.coef_

  # print(coeff)
  # print("std",x_std)
  # print("list",features_list)

  feat_imp_each_df = pd.DataFrame(columns=['Feature', 'Coeff'])
  feat_imp = pd.DataFrame(list(zip(features_list, coeff, x_std)), columns =['Feature', 'Coeff', 'STD'])
  feat_imp_each_df = feat_imp_each_df.append(feat_imp)
  feat_imp_each_df = feat_imp_each_df.sort_values(by='Coeff', ascending=False)
  feat_imp_each_df["Y_STD"] = y_std
  feat_imp_each_df["Stand_Coeff"] = feat_imp_each_df["Coeff"]*(feat_imp_each_df["STD"]/feat_imp_each_df["Y_STD"])

  del df_data, model, target_df

  return feat_imp_each_df
```

```python
def convert_float(df):
    for col in df.columns:
        try:
            df[col] = df[col].astype(float)
        except ValueError:
            print(f"Column '{col}' could not be converted to float and was left unchanged.")
    return df
```
